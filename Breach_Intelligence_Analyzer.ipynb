```python
from openai import OpenAI
from getpass import getpass
import os

os.environ["OPENAI_API_KEY"] = getpass("Paste your API key: ")
client = OpenAI()

password = getpass("Enter a password to analyze: ").strip()

if password == "":
  print("Invalid password, please try again")
else:
  print(f"Received a password of {len(password)} characters.")

def analyze_password(password):
  """Return an AI generated security audit for the given password."""
  prompt = f"""You are an elite cybersecurity auditor.

  Analyse the password below across all six of these dimensions
  1. strength
  2. predictability
  3. brute force risk
  4. dictionary attack risk
  5. human pattern weakness
  6. improvement advice

  Password: {password}

  Return a score from 1 to 100 where 100 is strongest.
  Explain your reasoning for small business owner with no technical background."""

  try:
      response = client.responses.create(model="gpt-5.6-luna", input=prompt)
      return response.output_text
  except Exception as e:
      return f"Analysis failed: {e}"

result = analyze_password("hunter2")
print(result)
```
